```python
import pandas as pd

# Load the dataset
df = pd.read_csv("/content/dataset.csv")

# Display basic information
print("Shape of the dataset:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Basic Dataset Exploration

print("="*50)
print("Dataset Shape:", df.shape)

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nTarget Distribution:")
print(df['anomaly'].value_counts())

print("\nTarget Distribution (%):")
print(df['anomaly'].value_counts(normalize=True)*100)

print("\nChannel Distribution:")
print(df['channel'].value_counts())

print("\nSummary Statistics:")
display(df.describe())
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Create figure
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# ----------------------------
# Plot 1: Anomaly Distribution
# ----------------------------
anomaly_counts = df['anomaly'].value_counts().sort_index()

axes[0].bar(['Normal (0)', 'Anomaly (1)'], anomaly_counts)

axes[0].set_title('Class Distribution')
axes[0].set_ylabel('Number of Samples')

for i, count in enumerate(anomaly_counts):
    axes[0].text(i, count + 20, str(count), ha='center', fontsize=10)

# ----------------------------
# Plot 2: Channel Distribution
# ----------------------------
channel_counts = df['channel'].value_counts()

axes[1].bar(channel_counts.index, channel_counts.values)

axes[1].set_title('Telemetry Channel Distribution')
axes[1].set_ylabel('Number of Samples')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder

# Create a copy of the dataset
df_processed = df.copy()

# Drop non-predictive columns
df_processed.drop(columns=['segment', 'train'], inplace=True)

# Encode the categorical 'channel' column
label_encoder = LabelEncoder()
df_processed['channel'] = label_encoder.fit_transform(df_processed['channel'])

# Separate features and target
X = df_processed.drop('anomaly', axis=1)
y = df_processed['anomaly']

print("Feature Matrix Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nEncoded Channel Mapping:")
for original, encoded in zip(label_encoder.classes_,
                             label_encoder.transform(label_encoder.classes_)):
    print(f"{original} --> {encoded}")

print("\nFirst 5 rows of processed data:")
display(df_processed.head())
```

```python
import matplotlib.pyplot as plt

# Compute correlation matrix
corr_matrix = X.corr()

# Plot heatmap
plt.figure(figsize=(16, 12))

im = plt.imshow(corr_matrix, cmap='coolwarm', aspect='auto')

plt.colorbar(im)

plt.xticks(range(len(corr_matrix.columns)),
           corr_matrix.columns,
           rotation=90,
           fontsize=8)

plt.yticks(range(len(corr_matrix.columns)),
           corr_matrix.columns,
           fontsize=8)

plt.title("Feature Correlation Heatmap")

plt.tight_layout()
plt.show()

# Find highly correlated features
threshold = 0.90

high_corr = []

columns = corr_matrix.columns

for i in range(len(columns)):
    for j in range(i):
        if abs(corr_matrix.iloc[i, j]) > threshold:
            high_corr.append((
                columns[i],
                columns[j],
                round(corr_matrix.iloc[i, j], 3)
            ))

print(f"\nHighly Correlated Feature Pairs (>|{threshold}|):")

for feature1, feature2, corr in high_corr:
    print(f"{feature1} <--> {feature2} : {corr}")
```

```python
# Remove highly correlated features

features_to_drop = [
    'std',
    'diff2_peaks',
    'diff2_var',
    'len_weighted',
    'var_div_duration'
]

X = X.drop(columns=features_to_drop)

print("Remaining Features:", X.shape[1])
print("\nFeature Names:\n")
print(X.columns.tolist())
```

```python
from sklearn.model_selection import train_test_split

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Set Shape:", X_train.shape)
print("Testing Set Shape :", X_test.shape)

print("\nTraining Class Distribution:")
print(y_train.value_counts())

print("\nTesting Class Distribution:")
print(y_test.value_counts())
```

```python
from sklearn.preprocessing import StandardScaler

# Initialize the scaler
scaler = StandardScaler()

# Scale only the training data
X_train_scaled = scaler.fit_transform(X_train)

# Use the same scaler on the test data
X_test_scaled = scaler.transform(X_test)

print("Scaling completed successfully!")

print("\nScaled Training Shape:", X_train_scaled.shape)
print("Scaled Testing Shape :", X_test_scaled.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

# Initialize the Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    random_state=42,
    class_weight='balanced'
)

# Train the model
rf_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Predictions
rf_pred = rf_model.predict(X_test)
rf_prob = rf_model.predict_proba(X_test)[:, 1]

# Evaluation
print("="*50)
print("Random Forest Performance")
print("="*50)

print(f"Accuracy : {accuracy_score(y_test, rf_pred):.4f}")
print(f"Precision: {precision_score(y_test, rf_pred):.4f}")
print(f"Recall   : {recall_score(y_test, rf_pred):.4f}")
print(f"F1 Score : {f1_score(y_test, rf_pred):.4f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, rf_prob):.4f}")

print("\nClassification Report")
print(classification_report(y_test, rf_pred))

print("\nConfusion Matrix")
print(confusion_matrix(y_test, rf_pred))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Feature Importance
importance = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': rf_model.feature_importances_
})

importance = importance.sort_values(by='Importance', ascending=False)

plt.figure(figsize=(10,6))

plt.barh(
    importance['Feature'],
    importance['Importance']
)

plt.xlabel("Importance Score")
plt.ylabel("Features")
plt.title("Random Forest Feature Importance")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()

display(importance)
```

```python
from sklearn.svm import SVC

# Initialize SVM
svm_model = SVC(
    kernel='rbf',
    C=1.0,
    gamma='scale',
    probability=True,
    class_weight='balanced',
    random_state=42
)

# Train
svm_model.fit(X_train_scaled, y_train)

print("SVM model trained successfully!")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Performance metrics
metrics = ['Accuracy', 'Precision', 'Recall', 'F1 Score', 'ROC-AUC']

rf = results.loc[0, metrics]
xgb = results.loc[1, metrics]
svm = results.loc[2, metrics]

x = np.arange(len(metrics))
width = 0.23

# Monochromatic palette
colors = plt.cm.Blues([0.85, 0.60, 0.35])

plt.figure(figsize=(10,6), dpi=300)

plt.bar(x-width, rf, width,
        color=colors[0],
        edgecolor='black',
        linewidth=0.6,
        label='Random Forest')

plt.bar(x, xgb, width,
        color=colors[1],
        edgecolor='black',
        linewidth=0.6,
        label='XGBoost')

plt.bar(x+width, svm, width,
        color=colors[2],
        edgecolor='black',
        linewidth=0.6,
        label='SVM')

plt.xticks(x, metrics, fontsize=11)
plt.yticks(fontsize=10)

plt.ylabel("Score", fontsize=12)
plt.xlabel("Evaluation Metric", fontsize=12)

plt.ylim(0.75, 1.01)

# Subtle horizontal grid
plt.grid(axis='y',
         linestyle='--',
         linewidth=0.5,
         alpha=0.4)

# Clean spines
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.legend(frameon=False)

plt.tight_layout()

plt.show()
```

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

# Compute ROC Curves
rf_fpr, rf_tpr, _ = roc_curve(y_test, rf_prob)
xgb_fpr, xgb_tpr, _ = roc_curve(y_test, xgb_prob)
svm_fpr, svm_tpr, _ = roc_curve(y_test, svm_prob)

# Compute AUC Scores
rf_auc = roc_auc_score(y_test, rf_prob)
xgb_auc = roc_auc_score(y_test, xgb_prob)
svm_auc = roc_auc_score(y_test, svm_prob)

# Publication-quality color palette (Tableau 10)
rf_color = "#4E79A7"      # Blue
xgb_color = "#F28E2B"     # Orange
svm_color = "#59A14F"     # Green
diag_color = "#A0A0A0"    # Grey

plt.figure(figsize=(8, 6), dpi=300)

# Plot ROC Curves
plt.plot(
    rf_fpr,
    rf_tpr,
    color=rf_color,
    linewidth=2.8,
    label=f"Random Forest (AUC = {rf_auc:.3f})"
)

plt.plot(
    xgb_fpr,
    xgb_tpr,
    color=xgb_color,
    linewidth=2.8,
    linestyle='--',
    label=f"XGBoost (AUC = {xgb_auc:.3f})"
)

plt.plot(
    svm_fpr,
    svm_tpr,
    color=svm_color,
    linewidth=2.8,
    linestyle='-.',
    label=f"SVM (AUC = {svm_auc:.3f})"
)

# Random Classifier
plt.plot(
    [0, 1],
    [0, 1],
    color=diag_color,
    linewidth=1.5,
    linestyle=':',
    label='Random Classifier'
)

# Labels
plt.xlabel("False Positive Rate", fontsize=12)
plt.ylabel("True Positive Rate", fontsize=12)
plt.title("Receiver Operating Characteristic (ROC) Curves",
          fontsize=13,
          pad=12)

# Grid
plt.grid(
    linestyle='--',
    linewidth=0.5,
    alpha=0.30
)

# Clean axes
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Tick font sizes
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

# Legend
plt.legend(
    loc='lower right',
    frameon=False,
    fontsize=10
)

plt.tight_layout()

plt.show()
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(15, 5), dpi=300)

models = [
    ("Random Forest", rf_pred),
    ("XGBoost", xgb_pred),
    ("Support Vector Machine", svm_pred)
]

for ax, (title, pred) in zip(axes, models):

    disp = ConfusionMatrixDisplay.from_predictions(
        y_test,
        pred,
        cmap="Blues",          # Gradient blue
        colorbar=False,
        values_format='d',
        ax=ax
    )

    ax.set_title(title, fontsize=12, pad=10)

    ax.set_xlabel("Predicted Label", fontsize=10)
    ax.set_ylabel("True Label", fontsize=10)

    # Make cells square
    ax.set_aspect('equal')

    # Improve annotation appearance
    for text in disp.text_.ravel():
        text.set_fontsize(13)
        text.set_fontweight('bold')

    # Remove unnecessary borders
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```
